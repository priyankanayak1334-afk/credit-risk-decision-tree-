```python
# Cell 1: Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Libraries successfully loaded!")
```

```python
# Cell 2: Data Generation and Train-Test Split
np.random.seed(42)
n_samples = 1500

# Generating core banking risk variables
data = {
    'Credit_Score': np.random.normal(670, 70, n_samples).round().clip(400, 850),
    'Income': np.random.normal(65000, 18000, n_samples).round().clip(20000, 150000),
    'Debt_to_Income_Ratio': np.random.uniform(0.1, 0.6, n_samples).round(2),
    'Loan_Amount': np.random.normal(30000, 12000, n_samples).round().clip(5000, 90000)
}

df = pd.DataFrame(data)

# Risk logic equation determining automated approval
score = (df['Credit_Score'] - 580)/100 + (df['Income']/50000) - (df['Debt_to_Income_Ratio'] * 3.0)
df['Approved'] = (score > 0.35).astype(int)

# Splitting features (X) and label targets (y)
X = df.drop(columns='Approved')
y = df['Approved']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print(f"Data Pipeline Ready: {X_train.shape[0]} Training items, {X_test.shape[0]} Testing items.")
```

```python
# Cell 3: Unconstrained Model Evaluation
overfit_model = DecisionTreeClassifier(random_state=42)
overfit_model.fit(X_train, y_train)

overfit_train_acc = accuracy_score(y_train, overfit_model.predict(X_train))
overfit_test_acc = accuracy_score(y_test, overfit_model.predict(X_test))

print(f"--- Unconstrained Baseline Tree ---")
print(f"Training Accuracy : {overfit_train_acc * 100:.2f}%")
print(f"Testing Accuracy  : {overfit_test_acc * 100:.2f}%")
print(f"Variance Gap      : {(overfit_train_acc - overfit_test_acc) * 100:.2f}%")
```

```python
# Cell 4: Enforcing Optimization Metrics
optimized_model = DecisionTreeClassifier(max_depth=3, min_samples_leaf=20, random_state=42)
optimized_model.fit(X_train, y_train)

opt_train_acc = accuracy_score(y_train, optimized_model.predict(X_train))
opt_test_acc = accuracy_score(y_test, optimized_model.predict(X_test))

print(f"--- Optimized / Pruned Tree ---")
print(f"Training Accuracy : {opt_train_acc * 100:.2f}%")
print(f"Testing Accuracy  : {opt_test_acc * 100:.2f}%")
print(f"New Variance Gap  : {(opt_train_acc - opt_test_acc) * 100:.2f}%")
```

```python
# Cell 5: Dynamic Decision Map Plotting
plt.figure(figsize=(18, 10), dpi=300)
plot_tree(
    optimized_model,
    feature_names=list(X.columns),
    class_names=['REJECTED', 'APPROVED'],
    filled=True,
    rounded=True,
    fontsize=11
)
plt.title("Production Credit Risk Decision Layout Model", fontsize=16, fontweight='bold', pad=20)
plt.show()
```

```python
# Cell 6: Analytical Importance Graphing
importances = optimized_model.feature_importances_
feature_imp_df = pd.DataFrame({'Feature': X.columns, 'Importance': importances}).sort_values(by='Importance', ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x='Importance', y='Feature', data=feature_imp_df, palette='viridis')
plt.title('Feature Metrics Controlling Credit Decisions', fontsize=14, fontweight='bold')
plt.xlabel('Relative Analytical Importance Value')
plt.ylabel('Evaluated Features')
plt.xlim(0, 1)
plt.tight_layout()
plt.show()

print(feature_imp_df.to_string(index=False))
```

```python
# Cell 7: Performance Analytics Matrix
y_pred = optimized_model.predict(X_test)

print("=== FINAL STRATEGIC TESTING CONFUSION MATRIX ===")
print(confusion_matrix(y_test, y_pred))
print("\n=== SYSTEM PERFORMANCE OVERVIEW REPORT ===")
print(classification_report(y_test, y_pred, target_names=['REJECTED', 'APPROVED']))
```
